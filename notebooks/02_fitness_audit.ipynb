```python
meta_file = Path("../data/raw/SRP376262_metadata.tsv")

meta = pd.read_csv(meta_file, sep="\t")

print(meta.shape)
print(meta.columns.tolist())
```

```python
for col in meta.columns:
    if any(
        key in col.lower()
        for key in ["title", "sample", "library", "strategy", "source"]
    ):
        print("\n", col)
        print(meta[col].dropna().astype(str).head(20).tolist())
```

```python
search_terms = [
    "growth",
    "day 16",
    "day16",
    "plasmid",
    "rpe1",
    "k562",
]

mask = pd.Series(False, index=meta.index)

for col in meta.columns:
    s = meta[col].astype(str)

    for term in search_terms:
        mask |= s.str.contains(
            term,
            case=False,
            na=False
        )

candidates = meta.loc[mask].copy()

print("Candidate rows:", len(candidates))
```

```python
interesting_cols = [
    c for c in meta.columns
    if any(
        x in c.lower()
        for x in [
            "run",
            "experiment",
            "sample",
            "title",
            "strategy",
            "source",
            "selection",
            "bases",
            "spots",
        ]
    )
]

display(candidates[interesting_cols].head(100))
```

```python
for col in [
    "library_strategy",
    "library_source",
    "library_selection",
]:
    print("\n", col)
    print(meta[col].value_counts(dropna=False))
```

```python
non_tx = meta[
    meta["library_source"]
    .astype(str)
    .str.upper()
    .ne("TRANSCRIPTOMIC")
].copy()

print("Non-transcriptomic runs:", len(non_tx))

display(
    non_tx[
        [
            "run_accession",
            "experiment_title",
            "experiment_desc",
            "library_strategy",
            "library_source",
            "library_selection",
            "run_total_spots",
            "run_total_bases",
            "run_alias",
        ]
    ].head(100)
)
```

```python
search_cols = [
    "experiment_title",
    "experiment_desc",
    "library_name",
    "run_alias",
]

pattern = (
    r"growth|screen|gdna|genomic|plasmid|"
    r"day[\s_-]?16|day[\s_-]?17|"
    r"initial|final|t0|tfinal"
)

mask = pd.Series(False, index=meta.index)

for col in search_cols:
    mask |= (
        meta[col]
        .astype(str)
        .str.contains(pattern, case=False, regex=True, na=False)
    )

growth_candidates = meta.loc[mask].copy()

print("Growth-screen candidates:", len(growth_candidates))

display(
    growth_candidates[
        [
            "run_accession",
            "experiment_title",
            "experiment_desc",
            "library_strategy",
            "library_source",
            "library_selection",
            "sample_accession",
            "run_total_spots",
            "run_total_bases",
            "run_alias",
        ]
    ].head(100)
)
```

# QC the two dual-guide replicates

```python
dual_cols = [
    "dual_rep1_gamma",
    "dual_rep2_gamma",
]

single_cols = [
    "single_rep1_gamma",
    "single_rep2_gamma",
]

print("Total rows:", len(k562_fit_raw))

print("\nMissing dual phenotypes:")
print(k562_fit_raw[dual_cols].isna().sum())

print("\nMissing single phenotypes:")
print(k562_fit_raw[single_cols].isna().sum())
```

```python
from scipy.stats import pearsonr, spearmanr

dual_complete = k562_fit_raw.dropna(
    subset=dual_cols
).copy()

pearson = pearsonr(
    dual_complete["dual_rep1_gamma"],
    dual_complete["dual_rep2_gamma"],
)

spearman = spearmanr(
    dual_complete["dual_rep1_gamma"],
    dual_complete["dual_rep2_gamma"],
)

print(f"Complete dual-guide rows: {len(dual_complete):,}")
print(f"Pearson r: {pearson.statistic:.3f}")
print(f"Spearman rho: {spearman.statistic:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))

plt.scatter(
    dual_complete["dual_rep1_gamma"],
    dual_complete["dual_rep2_gamma"],
    s=5,
    alpha=0.25,
)

plt.xlabel("Dual CRISPRi replicate 1 γ")
plt.ylabel("Dual CRISPRi replicate 2 γ")
plt.title("K562 CRISPRi growth-screen reproducibility")

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.tight_layout()
plt.show()
```

```python
k562_fitness = (
    k562_fit_raw
    .dropna(subset=dual_cols)
    [["gene", "sgID"] + dual_cols]
    .copy()
)

k562_fitness["k562_gamma"] = (
    k562_fitness[dual_cols].mean(axis=1)
)

k562_fitness.head()
```

```python
gene_counts = k562_fitness["gene"].value_counts()

print("Unique genes:", k562_fitness["gene"].nunique())
print("Genes appearing >1 time:", (gene_counts > 1).sum())

display(gene_counts.head(20))
```

```python
shared = pd.read_csv(
    "../results/phase0/shared_perturbations.csv"
)

shared_perts = set(
    shared["perturbation"].astype(str)
)

fitness_genes = set(
    k562_fitness["gene"]
    .dropna()
    .astype(str)
)

k562_overlap = sorted(
    shared_perts & fitness_genes
)

print("Shared K562/RPE1 Perturb-seq targets:", len(shared_perts))
print("Genes with K562 fitness:", len(fitness_genes))
print("Pilot overlap:", len(k562_overlap))
```

```python
k562_pilot_fitness = (
    k562_fitness[
        k562_fitness["gene"].isin(k562_overlap)
    ]
    .copy()
)

k562_pilot_fitness["dependency_strength"] = (
    -k562_pilot_fitness["k562_gamma"]
)

k562_pilot_fitness = (
    k562_pilot_fitness
    .sort_values(
        "dependency_strength",
        ascending=False
    )
    .reset_index(drop=True)
)

display(k562_pilot_fitness.head(20))
```

```python
plt.figure(figsize=(7, 4))

plt.hist(
    k562_pilot_fitness["k562_gamma"],
    bins=60
)

plt.xlabel("K562 growth phenotype γ")
plt.ylabel("Number of perturbations")
plt.title("K562 functional fitness distribution")

plt.tight_layout()
plt.show()
```

```python
RESULTS = Path("../results/phase0")
RESULTS.mkdir(parents=True, exist_ok=True)

k562_pilot_fitness.to_csv(
    RESULTS / "k562_fitness.csv",
    index=False
)
```

```python
k562_fitness_clean = (
    k562_fitness[
        k562_fitness["gene"] != "non-targeting"
    ]
    .copy()
)

print("Rows after removing controls:", len(k562_fitness_clean))
```

```python
k562_gene_fitness = (
    k562_fitness_clean
    .groupby("gene", as_index=False)
    .agg(
        k562_gamma=("k562_gamma", "median"),
        n_constructs=("sgID", "nunique"),
        rep1_gamma=("dual_rep1_gamma", "median"),
        rep2_gamma=("dual_rep2_gamma", "median"),
    )
)

print("Gene-level fitness rows:", len(k562_gene_fitness))
print()

print("Constructs per gene:")
print(k562_gene_fitness["n_constructs"].value_counts().sort_index())
```

```python
k562_gene_fitness["gene"].is_unique
```

```python
k562_pilot_fitness = (
    k562_gene_fitness[
        k562_gene_fitness["gene"].isin(shared_perts)
    ]
    .copy()
)

k562_pilot_fitness["dependency_strength"] = (
    -k562_pilot_fitness["k562_gamma"]
)

k562_pilot_fitness = (
    k562_pilot_fitness
    .sort_values(
        "dependency_strength",
        ascending=False
    )
    .reset_index(drop=True)
)

print("Final pilot genes:", len(k562_pilot_fitness))
print("Unique genes:", k562_pilot_fitness["gene"].nunique())

display(k562_pilot_fitness.head(10))
```

```python
plt.figure(figsize=(7, 4))

plt.hist(
    k562_pilot_fitness["k562_gamma"],
    bins=60
)

plt.xlabel("K562 growth phenotype γ")
plt.ylabel("Number of genes")
plt.title("K562 gene-level functional fitness distribution")

plt.tight_layout()
plt.show()
```
