```python
meta_file = Path("../data/raw/SRP376262_metadata.tsv")

meta = pd.read_csv(meta_file, sep="\t")

print(meta.shape)
print(meta.columns.tolist())
```

```python
for col in meta.columns:
    if any(
        key in col.lower()
        for key in ["title", "sample", "library", "strategy", "source"]
    ):
        print("\n", col)
        print(meta[col].dropna().astype(str).head(20).tolist())
```

```python
search_terms = [
    "growth",
    "day 16",
    "day16",
    "plasmid",
    "rpe1",
    "k562",
]

mask = pd.Series(False, index=meta.index)

for col in meta.columns:
    s = meta[col].astype(str)

    for term in search_terms:
        mask |= s.str.contains(
            term,
            case=False,
            na=False
        )

candidates = meta.loc[mask].copy()

print("Candidate rows:", len(candidates))
```

```python
interesting_cols = [
    c for c in meta.columns
    if any(
        x in c.lower()
        for x in [
            "run",
            "experiment",
            "sample",
            "title",
            "strategy",
            "source",
            "selection",
            "bases",
            "spots",
        ]
    )
]

display(candidates[interesting_cols].head(100))
```

```python
for col in [
    "library_strategy",
    "library_source",
    "library_selection",
]:
    print("\n", col)
    print(meta[col].value_counts(dropna=False))
```

```python
non_tx = meta[
    meta["library_source"]
    .astype(str)
    .str.upper()
    .ne("TRANSCRIPTOMIC")
].copy()

print("Non-transcriptomic runs:", len(non_tx))

display(
    non_tx[
        [
            "run_accession",
            "experiment_title",
            "experiment_desc",
            "library_strategy",
            "library_source",
            "library_selection",
            "run_total_spots",
            "run_total_bases",
            "run_alias",
        ]
    ].head(100)
)
```

```python
search_cols = [
    "experiment_title",
    "experiment_desc",
    "library_name",
    "run_alias",
]

pattern = (
    r"growth|screen|gdna|genomic|plasmid|"
    r"day[\s_-]?16|day[\s_-]?17|"
    r"initial|final|t0|tfinal"
)

mask = pd.Series(False, index=meta.index)

for col in search_cols:
    mask |= (
        meta[col]
        .astype(str)
        .str.contains(pattern, case=False, regex=True, na=False)
    )

growth_candidates = meta.loc[mask].copy()

print("Growth-screen candidates:", len(growth_candidates))

display(
    growth_candidates[
        [
            "run_accession",
            "experiment_title",
            "experiment_desc",
            "library_strategy",
            "library_source",
            "library_selection",
            "sample_accession",
            "run_total_spots",
            "run_total_bases",
            "run_alias",
        ]
    ].head(100)
)
```
